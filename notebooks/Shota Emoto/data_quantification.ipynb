```python
from pathlib import Path
import pandas as pd

# ---------------------------------
# Paths
# ---------------------------------

PROJECT_ROOT = Path.cwd().parents[1]

DATA_PATH = PROJECT_ROOT / "data" / "raw" / "cpdb_country_all.csv"

DOCS_PATH = PROJECT_ROOT / "docs" / "descriptive_statistics_CPDB.md"

# ---------------------------------
# Load data
# ---------------------------------

df = pd.read_csv(DATA_PATH)

print(df.shape)

display(df.head())
```

```python
# ---------------------------------
# Missing values
# ---------------------------------

missing_stats = pd.DataFrame({
    "missing_count": df.isnull().sum(),
    "missing_ratio": df.isnull().mean().round(3)
}).sort_values("missing_ratio", ascending=False)

# ---------------------------------
# Unique values
# ---------------------------------

unique_stats = pd.DataFrame({
    "n_unique": df.nunique()
}).sort_values("n_unique", ascending=False)

# ---------------------------------
# Policy counts
# ---------------------------------

n_policies = df["policy_id"].nunique()

n_countries = df["country_iso"].nunique()

# ---------------------------------
# Date coverage
# ---------------------------------

decision_dates = pd.to_datetime(
    df["decision_date"],
    errors="coerce"
)

start_dates = pd.to_datetime(
    df["start_date"],
    errors="coerce"
)

end_dates = pd.to_datetime(
    df["end_date"],
    errors="coerce"
)
```

```python
policy_instruments = (
    df["policy_instrument"]
    .dropna()
    .value_counts()
)

sectors = (
    df["sector"]
    .dropna()
    .value_counts()
)

policy_types = (
    df["policy_type"]
    .dropna()
    .value_counts()
)

policy_objectives = (
    df["policy_objective"]
    .dropna()
    .value_counts()
)

policy_status = (
    df["policy_status"]
    .dropna()
    .value_counts()
)
```

```python
markdown = f"""
# CPDB Descriptive Statistics

## Dataset Overview

- Number of observations: {len(df):,}
- Number of unique policies: {n_policies:,}
- Number of countries: {n_countries:,}

## Temporal Coverage

### Decision Date
- Earliest: {decision_dates.min()}
- Latest: {decision_dates.max()}

### Start Date
- Earliest: {start_dates.min()}
- Latest: {start_dates.max()}

### End Date
- Earliest: {end_dates.min()}
- Latest: {end_dates.max()}

---

# Missing Value Statistics

{missing_stats.to_markdown()}

---

# Number of Unique Values by Variable

{unique_stats.to_markdown()}

---

# Policy Instrument Distribution

{policy_instruments.head(30).to_markdown()}

---

# Sector Distribution

{sectors.head(30).to_markdown()}

---

# Policy Type Distribution

{policy_types.head(30).to_markdown()}

---

# Policy Objective Distribution

{policy_objectives.head(30).to_markdown()}

---

# Policy Status Distribution

{policy_status.to_markdown()}
"""
```

```python
DOCS_PATH.parent.mkdir(parents=True, exist_ok=True)

with open(DOCS_PATH, "w", encoding="utf-8") as f:
    f.write(markdown)

print(f"Saved to: {DOCS_PATH}")
```
