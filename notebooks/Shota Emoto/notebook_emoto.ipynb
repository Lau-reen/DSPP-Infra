```python
from pathlib import Path
import pandas as pd

from cpdb_api import request

DATA_DIR = Path("../data/temp")
DATA_DIR.mkdir(parents=True, exist_ok=True)

all_dfs = []

for country in countries:

    print(f"Downloading {country}...")

    r = request.Request()

    r.set_country(country)

    try:
        df = r.issue()

        if len(df) > 0:

            df["requested_country"] = country

            all_dfs.append(df)

            print(f"  success: {len(df)} rows")

        else:
            print("  empty")

    except Exception as e:

        print(f"  failed: {e}")
```

```python
combined_df = pd.concat(all_dfs, ignore_index=True)

print(combined_df.shape)

display(combined_df.head())

combined_df.to_csv(
    DATA_DIR / "cpdb_country_all.csv",
    index=False
)

print("Saved.")
```

```python
print(combined_df.columns.tolist())
```

```python
# 日付型に変換
combined_df["decision_date"] = pd.to_datetime(
    combined_df["decision_date"],
    errors="coerce"
)

# 集計
summary = (
    combined_df.groupby("requested_country")
    .agg(
        n_policies=("policy_id", "nunique"),
        first_year=("decision_date", "min"),
        last_year=("decision_date", "max"),
        n_objectives=("policy_objective", "nunique"),
        n_sectors=("sector", "nunique"),
        n_policy_instruments=("policy_instrument", "nunique"),
    )
    .reset_index()
)

display(summary.head())
```

```python
summary.to_csv(
    DATA_DIR / "cpdb_country_summary.csv",
    index=False
)

print("Saved.")
```
